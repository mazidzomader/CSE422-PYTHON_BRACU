## Question: A* Search Algorithm

Consider the following directed, weighted graph with nodes
A, B, C, D, E, F, G, H.

### Adjacency List (given)
* A : [(B, 4), (C, 2), (D, 7)]
* B : [(C, 1), (E, 3)]
* C : [(D, 2), (F, 8)]
* D : [(E, 2), (G, 5)]
* E : [(C, 6), (F, 1), (H, 4)]
* F : [(G, 3)]
* G : [(H, 2)]
* H : [(B, 5)]

### Heuristic Function h(n)

The heuristic values estimate the cost from each node to the goal node H.

| Node | A  | B | C | D | E | F | G | H |
| ---- | -- | - | - | - | - | - | - | - |
| h(n) | 10 | 8 | 7 | 6 | 4 | 3 | 1 | 0 |


### Tasks

Apply the A* search algorithm to find the shortest path from A to H.

For each iteration, show:

Current node selected

g(n) (path cost so far)

h(n) (heuristic value)

f(n) = g(n) + h(n)

List the order of node expansions.

Write the final optimal path and its total cost.

State whether the given heuristic is admissible, with justification.

Assumptions

A* uses f(n) = g(n) + h(n)

Ties in f(n) are broken by choosing the node with smaller h(n)

Graph is directed

````python
import heapq
import math
````

````python
def aStarSearch(graph, Start, End, hN):
    pq = []                      # priority queue (min-heap)
    closed = set()               # visited nodes

    # (f, g, node, path)
    heapq.heappush(pq, (hN[Start], 0, Start, [Start]))

    while pq:
        f, g, curr, path = heapq.heappop(pq)

        if curr in closed:
            continue
        closed.add(curr)

        if curr == End:
            return path, g       # path and total cost

        for neighbor, weight in graph[curr]:
            if neighbor not in closed:
                g_new = g + weight
                f_new = g_new + hN[neighbor]
                heapq.heappush(
                    pq,
                    (f_new, g_new, neighbor, path + [neighbor])
                )

    return None

graph = {
    'A': [('B', 4), ('C', 2), ('D', 7)],
    'B': [('C', 1), ('E', 3)],
    'C': [('D', 2), ('F', 8)],
    'D': [('E', 2), ('G', 5)],
    'E': [('C', 6), ('F', 1), ('H', 4), ('B', 1)],
    'F': [('G', 3)],
    'G': [('H', 2)],
    'H': [('B', 5)]
}

hN = {'A':10, 'B':8, 'C':7, 'D':6, 'E':4, 'F':3, 'G':1, 'H':0}

path, cost = aStarSearch(graph, 'A', 'H', hN)
print("Path:", path)
print("Cost:", cost)

````

Consider the following 5 × 5 grid.
Each cell represents a node. Movement is allowed only in four directions
(up, down, left, right). Each move has a cost of 1.

Grid Representation
```
S  .  .  X  .
.  X  .  X  .
.  X  .  .  .
.  .  X  X  .
X  .  .  .  G
```

Where:

- S = Start node
- G = Goal node
- X = Obstacle (cannot be traversed)
- . = Free cell
```
Start position: (0, 0)
Goal position: (4, 4)
```
Heuristic Function
```
Use Manhattan Distance:

h(n) = |x_goal − x_n| + |y_goal − y_n|
```
Tasks

- Model the grid as a state-space graph.

- Apply the A* search algorithm to find the shortest path from S to G.

- For each expanded node, compute:
```
    g(n) → path cost from S
    h(n) → heuristic value
    f(n) = g(n) + h(n)
```
- Show the order of node expansions.

- Draw the final path on the grid.

````python
import heapq

class Node:
    def __init__(self, position, g, h, parent=None, action=None):
        self.position = position
        self.g = g
        self.h = h
        self.f = g + h
        self.parent = parent
        self.action = action

    def __lt__(self, other):
        return self.f < other.f


def ManhattanDistance(x, y):
    return abs(x[0]-y[0]) + abs(x[1]-y[1])


def get_neighbors(position, maze, n, m):
    """Get valid neighboring positions and their corresponding actions"""
    row, col = position
    neighbors = []

    # Define moves: (delta_row, delta_col, action)
    moves = [
        (-1, 0, 'U'),  # Up
        (1, 0, 'D'),   # Down
        (0, -1, 'L'),  # Left
        (0, 1, 'R')    # Right
    ]

    for dr, dc, action in moves:
        new_row, new_col = row + dr, col + dc

        # Check if the new position is within bounds and is a valid path
        if 0 <= new_row < n and 0 <= new_col < m and maze[new_row][new_col] == '.':
            neighbors.append(((new_row, new_col), action))

    return neighbors

def reconstruct_path(node):
    """Reconstruct the path from start to goal"""
    path = []
    current = node

    while current.parent is not None:
        path.append(current.action)
        current = current.parent

    path.reverse()
    return len(path), ''.join(path)

def gridPathFinder(maze, start, goal, n, m):
    pq = []
    closed = set()

    startNode = Node(start, 0, ManhattanDistance(start, goal), None, "")

    heapq.heappush(pq, startNode)


    while pq:
        curr = heapq.heappop(pq)

        if curr.position == goal:
            return reconstruct_path(curr)

        if curr.position in closed:
            continue

        closed.add(curr.position)

        for neighbour, action in get_neighbors(curr.position, maze, n, m):
            if neighbour not in closed:
                g = curr.g + 1
                h = ManhattanDistance(neighbour, goal)

                neighNode = Node(neighbour, g, h, curr, action)

                heapq.heappush(pq, neighNode)

    return -1, "-1"

maze = [
    [".", ".", ".", "X", "."],
    [".", "X", ".", "X", "."],
    [".", "X", ".", ".", "."],
    [".", ".", "X", "X", "."],
    ["X", ".", ".", ".", "."]
]


print('\n'.join(['\t'.join([str(cell) for cell in row]) for row in maze]))

gridPathFinder(maze, (0,0), (4,4), 5, 5)
````

````python
def minimax(curD, idx, flag, val, maxD):
    if curD == maxD:
        return val[idx]

    if flag:
        return max(minimax(curD+1, idx*2, False, val, maxD), minimax(curD+1, idx*2+1, False, val, maxD))
    else:
        return min(minimax(curD+1, idx*2, True, val, maxD), minimax(curD+1, idx*2+1, True, val, maxD))

val = [3, 5, 2, 9, 12, 5, 23, 23]
import math
maxD = math.log(len(val), 2)

ans = minimax(0, 0, True, val, maxD)
print(ans)
````

````python
def alphaBeta(curD, idx, flag, val, maxD, a, b):
    if curD == maxD:
        return val[idx]

    if flag:
        left = alphaBeta(curD+1, idx*2, False, val, maxD, a, b)
        a = max(a, left)
        if a >= b:
            return left
        right = alphaBeta(curD+1, idx*2+1, False, val, maxD, a, b)
        return max(left, right)
    else:
        left = alphaBeta(curD+1, idx*2, True, val, maxD, a, b)
        b = min(b, left)
        if a >= b:
            return left
        right = alphaBeta(curD+1, idx*2+1, True, val, maxD, a, b)
        return min(left, right)

scores = [3, 5, 2, 9, 12, 5, 23, 23]

treeDepth = int(math.log(len(scores), 2))

print("The optimal value is : ", end = "")
print(alphaBeta(0, 0, True, scores, treeDepth, -math.inf, math.inf))
````
