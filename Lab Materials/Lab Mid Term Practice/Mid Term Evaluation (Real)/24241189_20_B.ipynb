```python
import random as rd
```

```python
def equation(a, b, c, d, x, y, z):
    eqn =  9*a + 1*b + 3*c + 7*d + 5*x + 6*y+ 3*z
    return eqn


def populationGeneration():
    pop = []

    for i in range(20):
        chrom = ""
        for j in range(7):
            chrom += str(rd.randint(0, 9))
        pop.append(chrom)

    return pop
def splitString(x):
    temp = []
    for i in range(len(x)):
        temp.append(int(x[i]))
    return temp
def fitnessCalc(chrom):
    n = len(str(chrom))
    for i in range(n):
        chrom = splitString(chrom)
        a = chrom[0]
        b = chrom[1]
        c = chrom[2]
        d = chrom[3]
        x = chrom[4]
        y = chrom[5]
        z = chrom[6]

        eqn = equation(a, b, c, d, x, y, z)
        fit = 1/(1+abs(eqn))

        return fit

def ParentSelection(arr):
    parent = []
    for i in range(4):
        idx = rd.randint(0, 19)
        parent.append(arr[idx])
    return parent


def twoPointCrossover(parent1, parent2):
    n = len(parent1)

    idx1 = rd.randint(0, n//2)
    idx2 = rd.randint(idx1, n-2)

    p1x = parent1[:idx1]
    p1y = parent1[idx1:idx2]
    p1z = parent1[idx2:]

    p2x = parent2[:idx1]
    p2y = parent2[idx1:idx2]
    p2z = parent2[idx2:]

    c1 = p1x + p2y + p1z
    c2 = p2x + p1y + p2z

    return c1, c2


def Stringsplit(x):
    temp = []
    for i in range(len(x)):
        temp.append((x[i]))
    return temp

def Mutation(x):
    n = len(x)
    idx1 = rd.randint(0, n//2)
    idx2 = rd.randint(idx1, n-1)

    temp = Stringsplit(x)

    temp[idx1], temp[idx2] = temp[idx2], temp[idx1]

    return "".join(temp)


def soln():
    encodedChr = populationGeneration()
    print("Encoded Chromosomes:", encodedChr)
    parents = ParentSelection(encodedChr)
    print("Parents:", parents)
    child = []
    for i in range(0, len(parents), 2):
        c1, c2 = twoPointCrossover(parents[i], parents[i+1])
        child.append(c1)
        child.append(c2)
    print("Child:", child)
    childFitness = []
    mutatedChild = []
    mChildFit = []
    for i in range(len(child)):
        childFitness.append(fitnessCalc(child[i]))

    idx = []
    for i in range(len(child)//2):
        idx.append(rd.randint(0, len(child)-1))
    # print(idx)
    for i in range(len(idx)):
        p = idx[i]
        # print(child[p])
        mutatedChild.append((child[p]))
    print("Child Fitness: ", childFitness)
    print("Muteted Child:", mutatedChild)

    for i in range(len(mutatedChild)):
        mChildFit.append(fitnessCalc(mutatedChild[i]))

    print("Muteted Child Fitness:", mChildFit)

for i in range(2000):
    print(f"=======================Iteration {i+1}=========================")
    soln()
```
